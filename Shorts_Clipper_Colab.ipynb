# Shorts Clipper (Colab, free)
YouTube link daalo, popularity score ke saath shorts pao.

**3 steps:** (1) Menu `Runtime > Change runtime type > T4 GPU` (tez chalega) (2) Neeche Step 1 ka ▶ dabao (3) Step 2 mein link daal ke ▶ dabao.

> Note: Colab par kabhi kabhi YouTube download block ho jata hai. Aisa ho to apne PC par `start.bat` wala tareeka use karo (README dekho).

```
#@title Step 1: Setup (sirf ek baar, 2-3 min)
REPO = "https://github.com/sahmad-creator/shorts-clipper"  #@param {type:"string"}
import os
if not os.path.exists("/content/clipper_repo"):
    !git clone -q {REPO} /content/clipper_repo
%cd /content/clipper_repo
!pip -q install -U yt-dlp faster-whisper imageio-ffmpeg flask numpy requests
print("Setup ho gaya")
```

```
#@title Step 2: Link daalo aur ▶ dabao
URL = ""  #@param {type:"string"}
TOP = 5  #@param {type:"slider", min:1, max:12, step:1}
LANG = "auto"  #@param ["auto","hi","en"]
WHISPER = "small"  #@param ["base","small","medium"]
CAPTIONS = True  #@param {type:"boolean"}
import os, sys, shutil
os.environ["WHISPER_MODEL"] = WHISPER
sys.path.insert(0, "/content/clipper_repo")
import pipeline
from IPython.display import Video, display, HTML
OUT = "/content/clipper_out"
shutil.rmtree(OUT, ignore_errors=True)
res = pipeline.run(URL, OUT, TOP, None if LANG == "auto" else LANG, CAPTIONS)
for c in res["clips"]:
    display(HTML(f"<h3>Score {c['score']} &nbsp; ({c['start']}s - {c['end']}s)</h3><p>{c['text'][:200]}</p>"))
    display(Video(f"{OUT}/{c['file']}", embed=True, width=270))
shutil.make_archive("/content/shorts", "zip", OUT)
from google.colab import files
files.download("/content/shorts.zip")
```
